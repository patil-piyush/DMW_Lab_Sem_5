**PROBLEM STATEMENT:**

KNN Classification
A dataset collected from hospital showing details of medical test reports with symptom values observed in the
patient’s and medical test either positive or negative.
Write a program to build k-NN classifier models. If k=3, find the class of the point (6, 6).



## **Step 1: Prepare the dataset**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier

# Create dataset
data = {
    'S1': [2, 4, 4, 4, 6, 6],
    'S2': [4, 6, 4, 2, 4, 2],
    'Class': ['Negative', 'Negative', 'Positive', 'Negative', 'Negative', 'Positive']
}

df = pd.DataFrame(data)
print("Dataset:")
print(df)
```

## **Step 2: Build KNN Classifier**

```python
X = df[['S1', 'S2']].values
y = df['Class'].values

k = 3
knn = KNeighborsClassifier(n_neighbors=k)
knn.fit(X, y)
```

## **Step 3: Predict for new point (6,6)**

```python
new_point = np.array([[6, 6]])
predicted_class = knn.predict(new_point)
print(f"\nPredicted class for point {new_point[0]}: {predicted_class[0]}")
```

## **Step 4: Visualization**

```python
plt.figure(figsize=(7, 5))

# Plot training points
for label, color in [('Positive', 'green'), ('Negative', 'red')]:
    subset = df[df['Class'] == label]
    plt.scatter(subset['S1'], subset['S2'], label=label, color=color, s=80, edgecolor='k')

# Plot new point
plt.scatter(new_point[0][0], new_point[0][1], color='blue', s=120, marker='*', label='Test Point (6,6)')

plt.title('k-NN Classification (k=3)')
plt.xlabel('Symptom S1')
plt.ylabel('Symptom S2')
plt.legend()
plt.grid(True)
plt.show()
```

## **Step 5: View nearest neighbors**

```python
distances, indices = knn.kneighbors(new_point)
print("\nNearest Neighbors (k=3):")
for i, index in enumerate(indices[0]):
    print(f"{i+1}) Point: {X[index]}, Class: {y[index]}, Distance: {distances[0][i]:.2f}")
```

## **Conclusion**

```python
print(f"\nConclusion: For k=3, the new patient with symptom values (6,6) "
      f"is classified as **{predicted_class[0]}**.")
```
